# Calculate RFM values

```python
import pandas as pd

reference_date = product_sales['InvoiceDate'].max() + pd.Timedelta(days=1)

rfm = (
    product_sales
    .dropna(subset=['Customer ID'])
    .groupby('Customer ID')
    .agg(
        Recency=('InvoiceDate', lambda x: (reference_date - x.max()).days),
        Frequency=('Invoice', 'nunique'),
        Monetary=('LineValue', 'sum')
    )
)

print("Reference date:", reference_date)
print(rfm.head(10).round(2))
```

# Score customers from 1 to 5

```python
rfm['R_Score'] = pd.qcut(
    rfm['Recency'].rank(method='first'),
    5,
    labels=[5, 4, 3, 2, 1]
).astype(int)

rfm['F_Score'] = pd.qcut(
    rfm['Frequency'].rank(method='first'),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

rfm['M_Score'] = pd.qcut(
    rfm['Monetary'].rank(method='first'),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

rfm['RFM_Score'] = (
    rfm['R_Score'].astype(str)
    + rfm['F_Score'].astype(str)
    + rfm['M_Score'].astype(str)
)

print(rfm.head(10))
```

# Assign customer segments

```python
def segment_customer(row):
    if row['R_Score'] >= 4 and row['F_Score'] >= 4 and row['M_Score'] >= 4:
        return 'Best Customers'
    elif row['R_Score'] >= 3 and row['F_Score'] >= 3:
        return 'Loyal Customers'
    elif row['R_Score'] >= 4:
        return 'Recent Customers'
    elif row['R_Score'] <= 2 and row['F_Score'] >= 3:
        return 'At Risk'
    else:
        return 'Other Customers'

rfm['Customer_Segment'] = rfm.apply(segment_customer, axis=1)

print(rfm[['Recency', 'Frequency', 'Monetary', 'RFM_Score', 'Customer_Segment']].head(10))
```

# Count customers in each segment

```python
segment_summary = (
    rfm['Customer_Segment']
    .value_counts()
    .rename_axis('Customer_Segment')
    .reset_index(name='Customer_Count')
)

segment_summary['Percentage'] = (
    segment_summary['Customer_Count']
    / segment_summary['Customer_Count'].sum()
    * 100
).round(2)

print(segment_summary)
```

# Visualize customer segments

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.bar(
    segment_summary['Customer_Segment'],
    segment_summary['Customer_Count']
)

plt.title('Customer Segmentation Using RFM Analysis')
plt.xlabel('Customer Segment')
plt.ylabel('Number of Customers')
plt.xticks(rotation=30, ha='right')

plt.tight_layout()
plt.show()
```

# Calculate sales for each RFM segment

```python
segment_sales = (
    rfm.groupby('Customer_Segment', observed=True)
    .agg(
        Customer_Count=('Monetary', 'size'),
        Total_Sales=('Monetary', 'sum'),
        Avg_Sales_Per_Customer=('Monetary', 'mean')
    )
    .sort_values('Total_Sales', ascending=False)
)

print(segment_sales.round(2))
```

# Recalculate ABC categories precisely

```python
import numpy as np

# Step 1: Sort products by sales
product_concentration = product_performance.sort_values(
    'Total_Sales', ascending=False
).copy()

# Step 2: Create product rank
product_concentration['Product_Rank'] = range(
    1, len(product_concentration) + 1
)

# Step 3: Calculate sales contribution
total_product_sales = product_concentration['Total_Sales'].sum()

product_concentration['Sales_Contribution_%'] = (
    product_concentration['Total_Sales'] / total_product_sales * 100
)

# Step 4: Calculate cumulative sales percentage
product_concentration['Cumulative_Sales_%'] = (
    product_concentration['Sales_Contribution_%'].cumsum()
)

# Step 5: Assign ABC categories
product_concentration['ABC_Category'] = np.select(
    [
        product_concentration['Cumulative_Sales_%'] <= 80,
        product_concentration['Cumulative_Sales_%'] <= 95
    ],
    ['A', 'B'],
    default='C'
)

# Step 6: Create ABC summary
abc_summary = (
    product_concentration.groupby('ABC_Category')
    .agg(
        Product_Count=('Product_Rank', 'count'),
        Total_Sales=('Total_Sales', 'sum')
    )
)

abc_summary['Product_Percentage_%'] = (
    abc_summary['Product_Count'] /
    len(product_concentration) * 100
).round(2)

abc_summary['Sales_Percentage_%'] = (
    abc_summary['Total_Sales'] /
    total_product_sales * 100
).round(2)

print(abc_summary.round(2))
```

# Calculate ABC product classification

```python
# ABC analysis based on product revenue

abc_products = (
    product_sales
    .groupby('StockCode')
    .agg(
        Product_Name=('Description', 'first'),
        Revenue=('LineValue', 'sum')
    )
    .reset_index()
    .sort_values('Revenue', ascending=False)
)

abc_products['Revenue_Contribution_%'] = (
    abc_products['Revenue']
    / abc_products['Revenue'].sum()
) * 100

abc_products['Cumulative_Revenue_%'] = (
    abc_products['Revenue_Contribution_%'].cumsum()
)

abc_products['ABC_Category'] = (
    abc_products['Cumulative_Revenue_%']
    .apply(lambda x: 'A' if x <= 80 else ('B' if x <= 95 else 'C'))
)

abc_summary = (
    abc_products.groupby('ABC_Category')
    .agg(
        Product_Count=('StockCode', 'nunique'),
        Revenue=('Revenue', 'sum')
    )
    .reset_index()
)

abc_summary['Revenue_Contribution_%'] = (
    abc_summary['Revenue'] / abc_products['Revenue'].sum() * 100
).round(2)

print(abc_summary.to_string(index=False))
```

# Calculate the next KPI — sales by weekday

```python
# Sales revenue by weekday

weekday_sales = (
    positive_sales
    .assign(
        Weekday=positive_sales['InvoiceDate'].dt.day_name()
    )
    .groupby('Weekday')
    .agg(
        Total_Sales=('LineValue', 'sum'),
        Unique_Orders=('Invoice', 'nunique')
    )
    .reset_index()
)

weekday_order = [
    'Monday', 'Tuesday', 'Wednesday',
    'Thursday', 'Friday', 'Saturday', 'Sunday'
]

weekday_sales['Weekday'] = pd.Categorical(
    weekday_sales['Weekday'],
    categories=weekday_order,
    ordered=True
)

weekday_sales = weekday_sales.sort_values('Weekday')

weekday_sales['Sales_Contribution_%'] = (
    weekday_sales['Total_Sales']
    / weekday_sales['Total_Sales'].sum()
    * 100
).round(2)

print(weekday_sales.to_string(index=False))
```
